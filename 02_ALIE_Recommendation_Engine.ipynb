```python
# ==========================================================================
# 2. SUBJECT CATALOG
# ==========================================================================

def build_subject_catalog():
    """One row per unique (Education_Level, Course, Subject) combination."""
    rows = []
    for entry in CURRICULUM:
        for subject in entry["subjects"]:
            rows.append({
                "Education_Level": entry["education_level"],
                "Course": entry["course"],
                "Stream": entry["stream"] if entry["stream"] else "General",
                "Subject": subject,
            })
    df = pd.DataFrame(rows).drop_duplicates().reset_index(drop=True)
    return df
```

```python
# ==========================================================================
# 3. TOPIC CATALOG (3,000+ rows)
# ==========================================================================

def build_topic_catalog(subject_catalog):
    """
    For every unique subject, expand its curated core concepts through
    TOPIC_TEMPLATES to build a beginner->advanced topic sequence with a
    logically chained Prerequisite field.
    """
    unique_subjects = sorted(subject_catalog["Subject"].unique())
    rows = []

    resource_by_difficulty = {
        "Beginner": ["Video", "PDF Notes", "Quiz"],
        "Intermediate": ["Quiz", "Practice Problems", "Video", "Case Study"],
        "Advanced": ["Interactive Lab", "Mini Project", "Case Study", "Practice Problems"],
    }
    time_by_difficulty = {"Beginner": (15, 30), "Intermediate": (30, 60), "Advanced": (60, 120)}
    skill_by_difficulty = {"Beginner": (1, 2), "Intermediate": (2, 4), "Advanced": (4, 5)}

    for subject in unique_subjects:
        concepts = core_topics_for(subject)
        previous_topic = None  # chains prerequisites across the whole subject
        for concept in concepts:
            for template, difficulty in TOPIC_TEMPLATES:
                topic_name = template.format(c=concept)
                lo, hi = time_by_difficulty[difficulty]
                est_time = int(np.random.randint(lo, hi + 1))
                lo_s, hi_s = skill_by_difficulty[difficulty]
                skill_level = int(np.random.randint(lo_s, hi_s + 1))
                resource = np.random.choice(resource_by_difficulty[difficulty])

                rows.append({
                    "Subject": subject,
                    "Topic": topic_name,
                    "Difficulty": difficulty,
                    "Estimated_Time": est_time,        # minutes
                    "Prerequisite": previous_topic if previous_topic else "None",
                    "Skill_Level": skill_level,          # 1 (novice) - 5 (expert)
                    "Recommended_Resource": resource,
                })
                previous_topic = topic_name

    df = pd.DataFrame(rows)
    # De-duplicate in the (rare) case two subjects share an identical topic name
    df = df.drop_duplicates(subset=["Subject", "Topic"]).reset_index(drop=True)
    return df
```

```python
# ==========================================================================
# 4. STUDENT PROFILES (~10,000 rows)
# ==========================================================================

def build_student_profiles(n_students=10000):
    rows = []
    # Weight courses so school-level education dominates enrollment, similar
    # to a real-world student population pyramid.
    level_weights = {"Grade 6-12": 0.45, "Diploma": 0.10, "Undergraduate": 0.35, "Postgraduate": 0.10}
    entries_by_level = {lvl: [e for e in CURRICULUM if e["education_level"] == lvl] for lvl in level_weights}

    levels = list(level_weights.keys())
    weights = list(level_weights.values())

    for i in range(1, n_students + 1):
        student_id = f"S{i:05d}"
        level = np.random.choice(levels, p=weights)
        entry = random.choice(entries_by_level[level])

        age_lo, age_hi = entry["age_range"]
        age = int(np.random.randint(age_lo, age_hi + 1))

        course_label = entry["course"]
        if entry["stream"] and "Grade" in entry["course"]:
            course_label = f'{entry["course"]} - {entry["stream"]}'

        if entry["term_type"] == "grade":
            semester_label = "Annual"
        elif entry["term_type"] == "year":
            semester_label = f"Year {np.random.randint(1, entry['num_terms'] + 1)}"
        else:
            semester_label = f"Semester {np.random.randint(1, entry['num_terms'] + 1)}"

        # A student registers for all subjects tied to their course/term
        # (mirrors a fixed-curriculum LMS enrollment), which anchors the
        # foreign-key relationship used by learning_history.
        registered_subjects = entry["subjects"]

        rows.append({
            "Student_ID": student_id,
            "Student_Name": generate_name(),
            "Age": age,
            "Education_Level": entry["education_level"],
            "Course": course_label,
            "Semester": semester_label,
            "Registered_Subjects": "; ".join(registered_subjects),
            "Learning_Goal": np.random.choice(LEARNING_GOALS),
        })

    df = pd.DataFrame(rows)
    assert df["Student_ID"].is_unique, "Duplicate Student_ID generated!"
    return df, entries_by_level
```

```python
# ==========================================================================
# 5. LEARNING HISTORY (~50,000 rows)
# ==========================================================================

def build_learning_history(student_df, topic_catalog, n_records=50000):
    """
    Simulates realistic study sessions:
      * A student only studies subjects they are registered in.
      * Topics within a subject are studied in curriculum sequence
        (mirrors prerequisite-respecting recommendations).
      * Engagement is skewed: a minority of (student, subject) pairs
        generate a disproportionate share of activity, like real LMS logs.
    """
    topics_by_subject = (
        topic_catalog.groupby("Subject")["Topic"].apply(list).to_dict()
    )
    est_time_lookup = dict(zip(
        zip(topic_catalog["Subject"], topic_catalog["Topic"]),
        topic_catalog["Estimated_Time"],
    ))

    # Build the full pool of valid (student, subject) pairs.
    pairs = []
    for _, srow in student_df.iterrows():
        student_id = srow["Student_ID"]
        subjects = srow["Registered_Subjects"].split("; ")
        for subj in subjects:
            if subj in topics_by_subject:  # guard against any catalog mismatch
                pairs.append((student_id, subj))

    n_pairs = len(pairs)
    # Skewed engagement weights (power-law-like) so some pairs are studied
    # much more intensively than others.
    weights = np.random.exponential(scale=1.0, size=n_pairs)
    weights = weights / weights.sum()

    sampled_idx = np.random.choice(n_pairs, size=n_records, p=weights, replace=True)

    # Count how many sessions each pair received, to assign sequence numbers.
    from collections import defaultdict
    session_counter = defaultdict(int)
    start_dates = {}
    rows = []

    today = datetime(2025, 6, 1)

    for idx in sampled_idx:
        student_id, subject = pairs[idx]
        session_counter[idx] += 1
        seq = session_counter[idx]

        topics = topics_by_subject[subject]
        n_topics = len(topics)

        current_topic = topics[(seq - 1) % n_topics]
        previous_topic = topics[(seq - 2) % n_topics] if seq > 1 else "None (Initial Assessment)"

        attempts = int(np.clip(np.round(np.random.exponential(1.4)) + 1, 1, 6))
        quiz_score = float(np.clip(np.random.normal(78 - 3 * (attempts - 1), 14), 0, 100))
        quiz_score = round(quiz_score, 1)

        base_time = est_time_lookup.get((subject, current_topic), 30)
        time_spent = int(np.clip(base_time * np.random.uniform(0.6, 1.9), 5, 240))

        topics_completed = min(seq, n_topics)
        learning_progress = round(topics_completed / n_topics, 2)

        if quiz_score >= 75:
            status = "Completed"
        elif quiz_score >= 40:
            status = "In Progress"
        else:
            status = "Needs Revision"

        if idx not in start_dates:
            start_dates[idx] = today - timedelta(days=int(np.random.randint(30, 540)))
        study_date = start_dates[idx] + timedelta(days=int(seq * np.random.randint(2, 12)))
        if study_date > today:
            study_date = today - timedelta(days=int(np.random.randint(0, 5)))

        rows.append({
            "Student_ID": student_id,
            "Subject": subject,
            "Topic": current_topic,
            "Quiz_Score": quiz_score,
            "Topics_Completed": topics_completed,
            "Time_Spent_Per_Topic": time_spent,          # minutes
            "Attempts": attempts,
            "Learning_Progress": learning_progress,       # 0-1
            "Previous_Recommendation": previous_topic,
            "Completion_Status": status,
            "Study_Date": study_date.strftime("%Y-%m-%d"),
        })

    df = pd.DataFrame(rows)
    return df
```

```python
# ==========================================================================
# 6. RESOURCE CATALOG
# ==========================================================================

def build_resource_catalog(topic_catalog):
    rows = []
    resource_name_bank = {
        "Video": ["Explainer Video", "Concept Walkthrough", "Lecture Recording"],
        "Quiz": ["Quick Check Quiz", "Chapter-End Quiz", "Adaptive Quiz"],
        "Practice Problems": ["Practice Problem Set", "Drill Exercises", "Worksheet"],
        "Interactive Lab": ["Virtual Lab", "Interactive Simulation", "Hands-on Lab"],
        "Case Study": ["Real-World Case Study", "Applied Case Analysis"],
        "Mini Project": ["Guided Mini Project", "Capstone Mini Project"],
        "PDF Notes": ["Concise Notes PDF", "Detailed Study Notes"],
    }

    for _, row in topic_catalog.iterrows():
        subject, topic, difficulty = row["Subject"], row["Topic"], row["Difficulty"]
        n_resources = np.random.randint(1, 3)  # 1-2 resources per topic
        chosen_types = np.random.choice(RESOURCE_TYPES, size=n_resources, replace=False)

        for r_type in chosen_types:
            base_name = np.random.choice(resource_name_bank[r_type])
            resource_name = f"{base_name}: {topic}"
            duration = int(np.random.randint(5, 90))
            rating = round(float(np.clip(np.random.normal(4.2, 0.5), 2.5, 5.0)), 1)
            url = f"https://lms.example.com/resources/{slugify(subject)}/{slugify(topic)}/{slugify(r_type)}"

            rows.append({
                "Topic": topic,
                "Resource_Name": resource_name,
                "Resource_Type": r_type,
                "Difficulty": difficulty,
                "Duration": duration,          # minutes
                "Rating": rating,               # 1-5 scale
                "URL_Placeholder": url,
            })

    return pd.DataFrame(rows)
```

```python
# ==========================================================================
# 7. VALIDATION (lightweight sanity checks before writing to disk)
# ==========================================================================

def validate(student_df, subject_df, topic_df, history_df, resource_df):
    errors = []

    if not student_df["Student_ID"].is_unique:
        errors.append("Duplicate Student_ID found in student_profiles.")

    valid_subjects = set(subject_df["Subject"])
    hist_subjects = set(history_df["Subject"])
    if not hist_subjects.issubset(valid_subjects):
        errors.append("learning_history references subjects not in subject_catalog.")

    valid_topics = set(zip(topic_df["Subject"], topic_df["Topic"]))
    hist_topics = set(zip(history_df["Subject"], history_df["Topic"]))
    if not hist_topics.issubset(valid_topics):
        errors.append("learning_history references (Subject, Topic) pairs not in topic_catalog.")

    valid_resource_topics = set(topic_df["Topic"])
    if not set(resource_df["Topic"]).issubset(valid_resource_topics):
        errors.append("resource_catalog references topics not in topic_catalog.")

    # Spot-check: every learning_history row's subject must be in that
    # student's Registered_Subjects.
    student_subjects = student_df.set_index("Student_ID")["Registered_Subjects"].to_dict()
    sample = history_df.sample(min(2000, len(history_df)), random_state=SEED)
    bad = sample[~sample.apply(
        lambda r: r["Subject"] in student_subjects.get(r["Student_ID"], "").split("; "), axis=1
    )]
    if len(bad) > 0:
        errors.append(f"{len(bad)} learning_history rows study a subject outside the student's registration.")

    if len(topic_df) < 3000:
        errors.append(f"topic_catalog has only {len(topic_df)} rows (< 3000 requirement).")

    if errors:
        raise AssertionError("Data validation failed:\n- " + "\n- ".join(errors))

    print("✔ All consistency checks passed.")
```

```python
# ==========================================================================
# 8. MAIN ORCHESTRATION
# ==========================================================================

def main(n_students=10000, n_history_records=50000):
    print("Building subject_catalog ...")
    subject_catalog = build_subject_catalog()

    print("Building topic_catalog ...")
    topic_catalog = build_topic_catalog(subject_catalog)

    print("Building student_profiles ...")
    student_profiles, _ = build_student_profiles(n_students)

    print("Building learning_history ...")
    learning_history = build_learning_history(student_profiles, topic_catalog, n_history_records)

    print("Building resource_catalog ...")
    resource_catalog = build_resource_catalog(topic_catalog)

    print("Validating referential integrity ...")
    validate(student_profiles, subject_catalog, topic_catalog, learning_history, resource_catalog)

    print("Saving CSV files ...")
    student_profiles.to_csv(os.path.join(OUTPUT_DIR, "student_profiles.csv"), index=False)
    subject_catalog.to_csv(os.path.join(OUTPUT_DIR, "subject_catalog.csv"), index=False)
    topic_catalog.to_csv(os.path.join(OUTPUT_DIR, "topic_catalog.csv"), index=False)
    learning_history.to_csv(os.path.join(OUTPUT_DIR, "learning_history.csv"), index=False)
    resource_catalog.to_csv(os.path.join(OUTPUT_DIR, "resource_catalog.csv"), index=False)

    print("\n===== SUMMARY =====")
    print(f"student_profiles.csv : {len(student_profiles):,} rows")
    print(f"subject_catalog.csv  : {len(subject_catalog):,} rows")
    print(f"topic_catalog.csv    : {len(topic_catalog):,} rows")
    print(f"learning_history.csv : {len(learning_history):,} rows")
    print(f"resource_catalog.csv : {len(resource_catalog):,} rows")
    print(f"\nAll files saved to ./{OUTPUT_DIR}/")


if __name__ == "__main__":
    main(n_students=10000, n_history_records=50000)
```

```python
def recommendation_confidence(
        quiz,
        progress,
        rating):

    confidence = (
        0.30*quiz +
        0.40*(progress*100) +
        0.30*(rating*20)
    )

    return round(min(confidence,100),2)
```

```python
def motivational_message(score):

    if score>=90:

        return (
            "Outstanding progress! "
            "You're ready to challenge yourself with advanced concepts."
        )

    elif score>=75:

        return (
            "Great consistency! "
            "Keep building on your strengths while improving weaker areas."
        )

    elif score>=60:

        return (
            "Good effort! "
            "Focus on completing prerequisite topics before moving ahead."
        )

    else:

        return (
            "Don't worry. Every expert started as a beginner. "
            "Let's strengthen the fundamentals step by step."
        )
```

# **THE HEART**

```python
def calculate_learning_health(student_id):

    history_df = get_learning_history(student_id)

    # No history available yet
    if history_df.empty:
        return 0.0

    avg_quiz = history_df["Quiz_Score"].fillna(0).mean()

    progress = history_df["Learning_Progress"].fillna(0).max() * 100

    avg_topics = history_df["Topics_Completed"].fillna(0).max()

    completion_rate = (
        (history_df["Completion_Status"] == "Completed").mean() * 100
    )

    avg_attempts = history_df["Attempts"].fillna(1).mean()

    topic_score = min((avg_topics / 10) * 100, 100)

    attempt_score = max(100 - (avg_attempts - 1) * 15, 40)

    health = (
        avg_quiz * 0.35
        + progress * 0.30
        + topic_score * 0.15
        + completion_rate * 0.10
        + attempt_score * 0.10
    )

    return round(min(100, health), 1)
```

```python
def generate_reason(subject,
                    quiz_score,
                    progress):

    reasons=[]

    if quiz_score<70:
        reasons.append(
            "Low quiz performance detected."
        )

    elif quiz_score<85:
        reasons.append(
            "Moderate understanding; strengthening this topic will improve consistency."
        )

    else:
        reasons.append(
            "Strong foundation. Ready for advanced learning."
        )

    if progress<50:
        reasons.append(
            "Learning progress is still developing."
        )

    else:
        reasons.append(
            "Prerequisite concepts have already been completed."
        )

    return reasons
```

```python
def expected_gain(difficulty):

    gains={
        "Beginner":12,
        "Intermediate":8,
        "Advanced":5
    }

    return gains[difficulty]
```

```python
def recommendation_priority(confidence):

    if confidence>=90:
        return "High"

    elif confidence>=75:
        return "Medium"

    else:
        return "Low"
```

```python
def generate_learning_path(student_id):

    # -----------------------------
    # Student Profile
    # -----------------------------
    student = get_student(student_id)

    if student is None:
        return "Student not found."

    # -----------------------------
    # Parse Registered Subjects
    # -----------------------------
    registered_subjects = parse_subjects(
        student["Registered_Subjects"]
    )

    # -----------------------------
    # Weak Subjects
    # -----------------------------
    weak_subjects = find_weak_subjects(student_id)

    recommendations = []

    # --------------------------------
    # Iterate from weakest subject first
    # --------------------------------
    for subject in weak_subjects.index:

        current_topic = latest_topic(student_id, subject)

        if current_topic is None:
            continue

        candidate_topics = next_topics(current_topic)

        if candidate_topics.empty:
            continue

        # Only Top Recommendation from this subject
        next_topic = candidate_topics.iloc[0]

        topic_name = next_topic["Topic"]

        resource = best_resource(topic_name)

        if resource is None:
            continue

        # Average Quiz Score
        history_df = get_learning_history(student_id)

        history_subject = history_df[
            history_df["Subject"] == subject
        ]

        avg_quiz = history_subject["Quiz_Score"].mean()

        avg_progress = history_subject[
            "Learning_Progress"
        ].mean()

        confidence = recommendation_confidence(
            avg_quiz,
            avg_progress,
            resource["Rating"]
        )

        recommendations.append({

    "Subject": subject,

    "Current Topic": current_topic,

    "Next Topic": topic_name,

    "Difficulty": next_topic["Difficulty"],

    "Estimated Time": next_topic["Estimated_Time"],

    "Resource": resource["Resource_Name"],

    "Resource Type": resource["Resource_Type"],

    "Confidence": confidence,

    "Priority":
        recommendation_priority(confidence),

    "Expected Improvement":
        expected_gain(next_topic["Difficulty"]),

    "Reason":
        generate_reason(
            subject,
            avg_quiz,
            avg_progress*100
        )
})

    recommendations = sorted(
        recommendations,
        key=lambda x: x["Confidence"],
        reverse=True
    )

    recommendations = recommendations[:3]

    learning_health = calculate_learning_health(student_id)

    output = {

        "Student":

        student["Student_Name"],

        "Course":

        student["Course"],

        "Semester":

        student["Semester"],

        "Goal":

        student["Learning_Goal"],

        "Learning Health":

        learning_health,

        "Motivation":

        motivational_message(
            learning_health
        ),

        "Recommendations":

        recommendations
    }

    return output
```

```python
def calculate_learning_streak(student_id):
    """
    Calculates the student's CURRENT learning streak: the number of
    consecutive calendar days (counting backward from their most recent
    study date) on which they logged at least one learning_history entry.

    Uses the existing Study_Date column only - no schema changes required.
    """
    history_df = get_learning_history(student_id)

    if history_df.empty:
        return 0

    # Unique study dates, sorted oldest -> newest
    study_dates = sorted(set(pd.to_datetime(history_df["Study_Date"]).dt.date))

    streak = 1
    for i in range(len(study_dates) - 1, 0, -1):
        gap_days = (study_dates[i] - study_dates[i - 1]).days
        if gap_days == 1:
            streak += 1
        else:
            break

    return streak

```

```python
# ==========================================================================
# 10. DAILY STUDY TARGET  &  11. RISK BADGE  (both derived from Learning Health)
# ==========================================================================

def daily_study_target_from_health(health):
    """
    Pure function: maps a Learning Health score (0-100) to a recommended
    daily study time target in minutes. Lower health -> higher target,
    to help the student close the gap faster; students who are already
    excelling get a lighter maintenance target.
    """
    if health >= 90:
        return 20
    elif health >= 75:
        return 30
    elif health >= 60:
        return 45
    else:
        return 60


def calculate_daily_study_target(student_id):
    """Standalone, student_id-based wrapper around daily_study_target_from_health()."""
    health = calculate_learning_health(student_id)
    return daily_study_target_from_health(health)


def risk_badge_from_health(health):
    """
    Pure function: maps a Learning Health score (0-100) to a Low /
    Moderate / High risk badge, flagging students who may need
    intervention before they fall further behind.
    """
    if health >= 75:
        return "Low"
    elif health >= 50:
        return "Moderate"
    else:
        return "High"


def calculate_risk_badge(student_id):
    """Standalone, student_id-based wrapper around risk_badge_from_health()."""
    health = calculate_learning_health(student_id)
    return risk_badge_from_health(health)
```

```python
# ==========================================================================
# 12. IMPROVED CONFIDENCE CALCULATION (normalized scoring)
# ==========================================================================

def recommendation_confidence_v2(quiz, progress, rating, attempts=1):
    """
    Normalized, more robust version of recommendation_confidence().
    Improvements over the original:
      - Every signal is individually clipped to a 0-100 scale BEFORE
        being combined, so a missing/out-of-range value can't distort
        the final score.
      - NaN-safe: missing quiz/progress/rating/attempts default sensibly
        instead of propagating NaN through the whole calculation.
      - Adds an Attempts signal: topics that took many attempts pull
        confidence down slightly, since that recommendation path hasn't
        been smooth for the student historically.
    Original recommendation_confidence() is untouched and still works.
    """
    quiz = 0 if pd.isna(quiz) else quiz
    progress = 0 if pd.isna(progress) else progress
    rating = 0 if pd.isna(rating) else rating
    attempts = 1 if pd.isna(attempts) else attempts

    quiz_norm = float(np.clip(quiz, 0, 100))
    progress_norm = float(np.clip(progress * 100, 0, 100))
    rating_norm = float(np.clip(rating * 20, 0, 100))
    attempt_penalty = float(np.clip((attempts - 1) * 5, 0, 20))

    confidence = (
        0.30 * quiz_norm
        + 0.35 * progress_norm
        + 0.25 * rating_norm
        + 0.10 * (100 - attempt_penalty)
    )

    return round(float(np.clip(confidence, 0, 100)), 2)
```

```python
# ==========================================================================
# 13. BETTER RECOMMENDATION REASONS
# ==========================================================================

def generate_reason_v2(subject, quiz_score, progress, attempts, difficulty):
    """
    Richer version of generate_reason(). Reuses the original quiz/progress
    reasoning (so behaviour doesn't regress) and layers on Attempts and
    target Difficulty, giving more specific, actionable explanations for
    why a topic was recommended.
    """
    reasons = generate_reason(subject, quiz_score, progress)  # reuse original logic

    if attempts is not None and not pd.isna(attempts):
        if attempts >= 4:
            reasons.append(
                "Multiple attempts on recent topics suggest extra practice "
                "will help before advancing further."
            )
        elif attempts <= 1:
            reasons.append(
                "Recent topics were cleared in a single attempt - a good "
                "sign for tackling this next."
            )

    has_quiz = quiz_score is not None and not pd.isna(quiz_score)
    if difficulty == "Advanced" and has_quiz and quiz_score >= 80:
        reasons.append("Strong recent scores make this a good time to attempt an advanced topic.")
    elif difficulty == "Beginner" and has_quiz and quiz_score < 60:
        reasons.append("Starting with a beginner-level topic will help rebuild confidence.")

    return reasons
```

```python
# ==========================================================================
# 14. ESTIMATED COMPLETION TIME
# ==========================================================================

def estimate_completion_time(student_id, subject):
    """
    Estimates how many days it will take the student to finish all
    remaining topics in `subject`, based on:
      - total topics for the subject (topic_catalog)
      - topics already completed (learning_history)
      - the student's historical study pace (sessions per active day)
    Returns None if there isn't enough history to estimate from.
    """
    subject_topics = get_topics(subject)
    total_topics = subject_topics["Topic"].nunique()

    if total_topics == 0:
        return None

    history_df = get_learning_history(student_id)
    subject_history = history_df[history_df["Subject"] == subject]

    if subject_history.empty:
        return None

    completed = subject_history["Topics_Completed"].max()
    remaining = max(total_topics - completed, 0)

    if remaining == 0:
        return 0

    active_days = subject_history["Study_Date"].nunique()
    sessions = len(subject_history)
    pace = (sessions / active_days) if active_days > 0 else 1  # sessions per active day
    pace = pace if pace > 0 else 1

    estimated_days = int(np.ceil(remaining / pace))
    return estimated_days
```

```python
# ==========================================================================
# 15. LEARNING PERSONA
# ==========================================================================

def determine_learning_persona(student_id):
    """
    Classifies the student into one of several learning personas based on
    their aggregate learning_history behaviour. Checked in priority order
    so the most distinctive pattern wins when a student fits more than one.
    """
    history_df = get_learning_history(student_id)

    if history_df.empty:
        return "New Learner"

    avg_quiz = history_df["Quiz_Score"].fillna(0).mean()
    avg_attempts = history_df["Attempts"].fillna(1).mean()
    n_subjects = history_df["Subject"].nunique()
    completion_rate = (history_df["Completion_Status"] == "Completed").mean() * 100
    streak = calculate_learning_streak(student_id)

    if avg_quiz >= 85 and completion_rate >= 60:
        return "High Performer"
    elif avg_attempts <= 1.5 and avg_quiz >= 75:
        return "Fast Learner"
    elif avg_attempts >= 3:
        return "Revision Seeker"
    elif streak >= 5:
        return "Consistent Learner"
    elif n_subjects >= 4:
        return "Explorer"
    else:
        return "Developing Learner"


```

```python
# ==========================================================================
# 16. ALIE SCORE (with category-wise breakdown)
# ==========================================================================

def calculate_alie_score(student_id):
    """
    ALIE Score: a holistic 0-100 composite score with a category-wise
    breakdown. Extends the signals already used in calculate_learning_health()
    with a Consistency (streak-based) signal for a fuller picture of the
    student's overall standing. Does not modify calculate_learning_health().
    """
    history_df = get_learning_history(student_id)

    if history_df.empty:
        return {"ALIE_Score": 0.0, "Breakdown": {}}

    performance = history_df["Quiz_Score"].fillna(0).mean()
    progress = history_df["Learning_Progress"].fillna(0).max() * 100
    completion_rate = (history_df["Completion_Status"] == "Completed").mean() * 100
    engagement = min((history_df["Topics_Completed"].fillna(0).max() / 10) * 100, 100)
    consistency = min(calculate_learning_streak(student_id) * 10, 100)

    breakdown = {
        "Performance": round(performance, 1),
        "Progress": round(progress, 1),
        "Completion": round(completion_rate, 1),
        "Engagement": round(engagement, 1),
        "Consistency": round(consistency, 1),
    }

    alie_score = (
        performance * 0.30
        + progress * 0.25
        + completion_rate * 0.15
        + engagement * 0.15
        + consistency * 0.15
    )

    return {
        "ALIE_Score": round(min(alie_score, 100), 1),
        "Breakdown": breakdown,
    }
```

```python
# ==========================================================================
# 17. ACHIEVEMENT BADGES
# ==========================================================================

def calculate_achievement_badges(student_id):
    """
    Returns a list of achievement badges (display strings) earned by the
    student, based on thresholds over their learning_history and derived
    ALIE metrics. Purely additive - reads existing data, changes nothing.
    """
    history_df = get_learning_history(student_id)

    if history_df.empty:
        return []

    badges = []

    streak = calculate_learning_streak(student_id)
    if streak >= 7:
        badges.append("7-Day Streak")
    elif streak >= 3:
        badges.append("3-Day Streak")

    if (history_df["Quiz_Score"] >= 90).any():
        badges.append("Quiz Ace")

    if history_df["Subject"].nunique() >= 4:
        badges.append("Subject Explorer")

    completed_count = (history_df["Completion_Status"] == "Completed").sum()
    if completed_count >= 20:
        badges.append("20 Topics Completed")
    elif completed_count >= 10:
        badges.append("10 Topics Completed")

    if (history_df["Attempts"] == 1).mean() >= 0.5:
        badges.append("First-Try Champion")

    if calculate_learning_health(student_id) >= 90:
        badges.append("Top Performer")

    return badges
```

```python
# ==========================================================================
# 18. ENHANCED LEARNING PATH  (Streamlit-ready wrapper)
# ==========================================================================

def generate_learning_path_v2(student_id):
    """
    Streamlit-ready enhanced learning path. Wraps the ORIGINAL
    generate_learning_path() unchanged (it remains fully usable on its
    own) and layers on all the new ALIE features on top:
      - Learning_Streak, Daily_Study_Target_Minutes, Risk_Badge
      - Learning_Persona, Achievement_Badges
      - ALIE_Score + ALIE_Score_Breakdown
      - each recommendation also gets Confidence_v2, Reason_v2 and
        Estimated_Completion_Days
    All original keys/values from generate_learning_path() are preserved.
    """
    base = generate_learning_path(student_id)

    if isinstance(base, str):  # "Student not found."
        return base

    health = base["Learning Health"]
    history_df = get_learning_history(student_id)

    enhanced_recommendations = []
    for rec in base["Recommendations"]:
        subject_history = history_df[history_df["Subject"] == rec["Subject"]]
        avg_quiz = subject_history["Quiz_Score"].mean() if not subject_history.empty else None
        avg_progress = subject_history["Learning_Progress"].mean() if not subject_history.empty else None
        avg_attempts = subject_history["Attempts"].mean() if not subject_history.empty else 1

        resource = best_resource(rec["Next Topic"])
        rating = resource["Rating"] if resource is not None else 0

        rec_v2 = dict(rec)  # keep every original field intact
        rec_v2["Confidence_v2"] = recommendation_confidence_v2(
            avg_quiz, avg_progress, rating, avg_attempts
        )
        rec_v2["Reason_v2"] = generate_reason_v2(
            rec["Subject"],
            avg_quiz,
            (avg_progress * 100) if avg_progress is not None else None,
            avg_attempts,
            rec["Difficulty"],
        )
        rec_v2["Estimated_Completion_Days"] = estimate_completion_time(student_id, rec["Subject"])

        enhanced_recommendations.append(rec_v2)

    alie = calculate_alie_score(student_id)

    enhanced_output = dict(base)  # preserve all original top-level keys
    enhanced_output["Recommendations"] = enhanced_recommendations
    enhanced_output["Learning_Streak"] = calculate_learning_streak(student_id)
    enhanced_output["Daily_Study_Target_Minutes"] = daily_study_target_from_health(health)
    enhanced_output["Risk_Badge"] = risk_badge_from_health(health)
    enhanced_output["Learning_Persona"] = determine_learning_persona(student_id)
    enhanced_output["Achievement_Badges"] = calculate_achievement_badges(student_id)
    enhanced_output["ALIE_Score"] = alie["ALIE_Score"]
    enhanced_output["ALIE_Score_Breakdown"] = alie["Breakdown"]

    return enhanced_output
```
